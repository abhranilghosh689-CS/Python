```python
# Match Case
x = int(input("Enter the value of number: "))
match x:
  case 0:
    print("Number is 0")
  case 4:
    print("Number is 4")
  case _ if x != 50:
    print(x, "is not 50")
  case _:
    print(x)
```

```python
# Check is a num is prime or not
def isPrime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
k = int(input("Enter: "))
if isPrime(k):
    print("Prime")
else:
    print("Not Prime")
```

```python
# Sum of digits of a num
def isSum(n):
    total = 0
    while n > 0:
        m = n % 10
        total = total + m
        n = n // 10
    print(total)
k = int(input("Enter: "))
isSum(k)
```

```python
# Printing Average using tuple
def average(*numbers):
  sum = 0
  for i in numbers:
    sum = sum + i
  a = sum / len(numbers)
  print(a)
average(9, 65, 23)
```
