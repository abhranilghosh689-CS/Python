```python
# KBC Game
def kbc_game():
    questions = [
        ["What is the capital of India?", "A. Mumbai", "B. Delhi", "C. Kolkata", "D. Chennai", "B"],
        ["Which planet is known as the Red Planet?", "A. Earth", "B. Venus", "C. Mars", "D. Jupiter", "C"],
        ["Who wrote 'Romeo and Juliet'?", "A. Shakespeare", "B. Dickens", "C. Austen", "D. Twain", "A"],
        ["What is 5 * 6?", "A. 11", "B. 30", "C. 56", "D. 20", "B"],
        ["Which is the largest ocean?", "A. Atlantic", "B. Indian", "C. Arctic", "D. Pacific", "D"]
    ]
    money = [1000, 5000, 10000, 50000, 100000]
    winnings = 0
    print(" Welcome to KBC Game! \n")
    for i in range(len(questions)):
        q = questions[i]
        print(f"\nQuestion for ₹{money[i]}:")
        print(q[0])
        print(q[1])
        print(q[2])
        print(q[3])
        print(q[4])
        answer = input("Enter your answer (A/B/C/D): ").upper()
        if answer == q[5]:
            print(" Correct answer!")
            winnings = money[i]
            print(f"You won ₹{winnings}")
        else:
            print(" Wrong answer!")
            print(f"Correct answer was {q[5]}")
            break
    print(f"\n Game Over! Total winnings: ₹{winnings}")
# Run the game
kbc_game()
```

```python
# Recursion
def factorial(n) :
  if n == 0 or n == 1 :
    return 1
  else:
    return n * factorial(n - 1)
print(factorial(int(input("Enter n: "))))
```

```python
# Fibonacci Series
def fibonacci(k) :
  if k == 0 :
    return 0
  elif k == 1 :
    return 1
  else :
    return fibonacci(k - 1) + fibonacci(k - 2)
print(fibonacci(int(input("Enter n: "))))
```

```python
# Counting vowels, consonants, upper_cases, lower_cases
s = input("Enter string: ") # String input

v = 0 # Counters set to 0
c = 0
u = 0
l = 0

vowel_set = "aeiouAEIOU" # Definining vowel set
for ch in s : # Conditions
  if ch.isalpha() :
    if ch in vowel_set :
      v += 1
    else :
      c += 1
  if ch.isupper() :
    u += 1
  elif ch.islower() :
    l += 1
print("Vowels: ",v) # Printing result
print("Consonants: ",c)
print("Upper_case: ",u)
print("Lower_case: ",l)
```
